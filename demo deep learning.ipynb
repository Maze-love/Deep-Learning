```python
def relu(x):
    return max(0, x)


def neuron(x1, x2, w1, w2, bias):
    # Step 1: weighted sum
    z = x1 * w1 + x2 * w2 + bias
    # Step 2: activation
    output = relu(z)
    # Step 3: return
    return output


output = neuron(
    2, 3,
    0.5, 0.2,
    1
)
print(output)
```

Part 8: Build the Complete Network

```python
# =========================
# Activation Function
# =========================
def relu(x):
    return max(0, x)
    
# =========================
# Neuron
# =========================
def neuron(x1, x2, w1, w2, bias):
    z = x1 * w1 + x2 * w2 + bias
    return relu(z)
# =========================
# Input
# =========================
x1 = 2
x2 = 3
# =========================
# Hidden Layer
# =========================
h1 = neuron(
    x1, x2,
    0.5, 0.2,
    1
)
h2 = neuron(
    x1, x2,
    -0.3, 0.8,
    0.5
)
# =========================
# Output Layer
# =========================
output = neuron(
    h1, h2,
    0.4, 0.6,
    0.2
)

# =========================
# Results
# =========================
print("Input 1:", x1)
print("Input 2:", x2)
print("Hidden neuron 1:", h1)
print("Hidden neuron 2:", h2)
print("Final output:", output)
```

Part 14: Complete 3-Neuron Network

```python
# =========================
# Activation Function
# =========================
def relu(x):
    return max(0, x)

# =========================
# Two-input Neuron
# =========================

def neuron(x1, x2, w1, w2, bias):
    z = x1 * w1 + x2 * w2 + bias
    return relu(z)

# =========================
# Output Neuron
# =========================
def output_neuron(h1, h2, h3, w1, w2, w3, bias):
    z = (
    h1 * w1
    + h2 * w2
    + h3 * w3
    + bias
    )
    return relu(z)

# =========================
# Input
# =========================
x1 = 2
x2 = 3

# =========================
# Hidden Layer
# =========================
h1 = neuron(
    x1, x2,
    0.5, 0.2,
    1
)

h2 = neuron(
    x1, x2,
    -0.3, 0.8,
    0.5
)

h3 = neuron(
    x1, x2,
    0.7, -0.4,
    0.3
)
# =========================
# Output Layer
# =========================
prediction = output_neuron(
    h1,
    h2,
    h3,
    0.4,
    0.6,
    0.2,
    0.1
)
# =========================
# Print Results
# =========================
print("Input 1:", x1)
print("Input 2:", x2)
print()
print("Hidden Layer:")
print("h1:", h1)
print("h2:", h2)
print("h3:", h3)
print()
print("Final Prediction:", prediction)
```
