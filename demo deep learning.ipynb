Part 8: Build the Complete Network

```python
# =========================
# Activation Function
# =========================
def relu(x):
    return max(0, x)
    
# =========================
# Neuron
# =========================
def neuron(x1, x2, w1, w2, bias):
    z = x1 * w1 + x2 * w2 + bias
    return relu(z)
# =========================
# Input
# =========================
x1 = 2
x2 = 3
# =========================
# Hidden Layer
# =========================
h1 = neuron(
    x1, x2,
    0.5, 0.2,
    1
)
h2 = neuron(
    x1, x2,
    -0.3, 0.8,
    0.5
)
# =========================
# Output Layer
# =========================
output = neuron(
    h1, h2,
    0.4, 0.6,
    0.2
)

# =========================
# Results
# =========================
print("Input 1:", x1)
print("Input 2:", x2)
print("Hidden neuron 1:", h1)
print("Hidden neuron 2:", h2)
print("Final output:", output)
```
